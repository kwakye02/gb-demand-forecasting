```python
# import necessary libraries
import pandas as pd
from pathlib import  Path
import holidays
from gb_demand_forecasting.features.lags import build_features, FEATURE_COLS

pd.set_option("display.max_columns", None)
PROJECT_ROOT = Path.cwd().parent
DATA_DIR = PROJECT_ROOT / "data" / "processed"
```

```python
#load the datasets
demand_path = DATA_DIR / "gb_demand_merged.parquet"
data = pd.read_parquet(demand_path)
df = data.copy()
```

```python
df = build_features(df)
df["seasonal_naive"] = df["lag_7d"]
df[FEATURE_COLS].isna().sum()
```

```python
df = df[df["SETTLEMENT_DATE"] >= "2020-01-01"].dropna(subset=FEATURE_COLS)
df.to_parquet(DATA_DIR / "gb_demand_features.parquet", index=True)
```
